```python
import os
import pandas as pd

# Directorio base donde tienes organizadas tus carpetas
BASE_DIR = './datos/'

ARCHIVOS = {
    'Módulo 1638 (Anemia)': '1638_anemia/RECH6_2025.csv',
    'Módulo 1629 (Hogar)': '1629_hogar/RECH1_2025.csv',
    'Módulo 1630 (Vivienda)': '1630_vivienda/RECH23_2025.csv',
    'Módulo 1633 (Embarazo)': '1633_embarazo/REC41_2025.csv'
}

# Inspección automática
for nombre_modulo, ruta_relativa in ARCHIVOS.items():
    ruta_completa = os.path.join(BASE_DIR, ruta_relativa)
    if os.path.exists(ruta_completa):
        df = pd.read_csv(ruta_completa, nrows=5)  # Carga rápida de solo 5 filas
        print("="*60)
        print(f"📌 {nombre_modulo}")
        print(f"Ruta: {ruta_completa}")
        print(f"Número de columnas: {len(df.columns)}")
        print("Lista de columnas encontradas:")
        print(list(df.columns))
    else:
        print(f"⚠️ No se encontró el archivo en: {ruta_completa}")
```
